```python

def transform(x):
    if pd.isnull(x[6]):
        return x
    # print(x[6])
    a = str(x[6]).lower()
    if 'built up area' in a:
        p = a.split('built up area')
        #p[0]- either empty or point to prev , always take p[1]
        val = re.findall(r'\d+\.?\d+', p[1])
        if '(' in p:
            k = p.split('(')
            val = re.findall(r'\d+\.?\d+', k[1])
            x[9] = givevalue(val,'m')
        else:
            val = re.findall(r'\d+\.?\d+', p[1])
            x[9] = givevalue(val,'y')
    else:
        x[9] = 0

    if 'super built up area' in a:
        p = a.split('super built up area')
        val = re.findall(r'\d+\.?\d+', p[1])
        if '(' in p:
            k = p.split('(')
            val = re.findall(r'\d+\.?\d+', k[1])
            x[8] = givevalue(val,'m')
        else:
            val = re.findall(r'\d+\.?\d+', p[1])
            x[8] = givevalue(val,'y')
    else:
        x[8] = 0
        
        
    if 'carpet area' in a:
        p = a.split('carpet area')
        val = re.findall(r'\d+\.?\d+', p[1])
        if '(' in p:
            k = p.split('(')
            val = re.findall(r'\d+\.?\d+', k[1])
            x[10] = givevalue(val,'m')
        else:
            val = re.findall(r'\d+\.?\d+', p[1])
            x[10] = givevalue(val,'y')
    else:
        x[10] = 0
        
    if 'plot area' in a:
        p = a.split('plot area')
        val = re.findall(r'\d+\.?\d+', p[1])
        if '(' in p:
            k = p.split('(')
            val = re.findall(r'\d+\.?\d+', k[1])
            x[7] = givevalue(val,'m')
        else:
            val = re.findall(r'\d+\.?\d+', p[1])
            x[7] = givevalue(val,'y')
    else:
        x[7] = 0;

    return x
        
    
        
def givevalue(val,compare):
    # print(val)
    val = float(val[0])
    # print(val)
    #compare is in sq.yards
    if compare=='y':
        # print("square yard")
        #val is in sq.yards
        return val*0.83*10.76
    elif compare=='m':
        # print("square meter")
        return val*10.76
    else:
        # print("square foot")
        return val;
        
    
```

```python
# new property-> 0-1 yr or less, 
#relatively new-> 1-5
#moderately old -> 5-10
#  old property ->10+
#undefined-undefined
##under construction
def categorize_age_possession(value):
    if pd.isna(value) or value=='undefined':
        return 'Undefined'
    elif value=='1 to 5 Year Old':
        return 'new Property'
    elif value=='5 to 10 Year Old':
        return 'Relatively new'
    elif value=='10+ Year Old':
        return 'old Property'
    elif 'Under Construction' in value or 'By' in value:
        return 'Under Construction'
    try:
        #having date
        value.split(" ")[-1]
        return 'Under Construction'
    except:
        return 'Undefined'
    
```

```python
dt['agePossession'] = dt['agePossession'].apply(categorize_age_possession)
```

```python
import ast
import re
import pandas as pd

k = set()

def furnish(val):
    if isinstance(val, str):
        try:
            val = ast.literal_eval(val)
        except (ValueError, SyntaxError):
            return

    if not isinstance(val, (list, tuple)):
        return

    for i in val:
        match = re.match(r'^\d+\d*\s+', i)

        if match:
            item = re.sub(r'^\d+\d*\s+', '', i)
        else:
            item = re.sub(r'^No\s+', '', i)

        k.add(item.strip())

# Test with your list
furnish(col)


# Process the DataFrame column
dt['furnishDetails'].apply(furnish)
print(k)
```

```python
col = ['3 Wardrobe', '1 Water Purifier', '7 Fan', '1 Exhaust Fan', '3 Geyser', '12 Light', '1 Modular Kitchen', '1 Chimney', 'No AC', 'No Bed',
       'No Curtains', 'No Dining Table', 'No Microwave', 'No Fridge', 'No Sofa', 'No Stove', 'No TV', 'No Washing Machine']

def transform(val,index):
    if isinstance(val, str):
        try:
            val = ast.literal_eval(val)
        except (ValueError, SyntaxError):
            return

    if not isinstance(val, (list, tuple)):
        return
    
    for i in val:
        match = re.match(r'^\d+\d*\s+', i)
        
        if match:
            num = match.group()
            item = re.sub(r'^\d+\d*\s+', '', i)
            # print(item,num)
            num = int(num)
        else:
            item = re.sub(r'^No\s+', '', i)
            # print(item,0)
            num = 0
        dt.loc[index,item]=num
            
 # Test with your list
transform(col,0)           
    
```

#### K means clustering

```python
col=[]
for i in k:
    col.append(i)
furnishing = dt[col]
furnishing.info()
```

```python
for i in furnishing.columns:
    print(furnishing[i].value_counts())
```

```python
from sklearn.cluster import KMeans
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()
scaled_data = scaler.fit_transform(furnishing)

wcss_reduced = []
for i in range(1,19):
    model = KMeans(n_clusters=i,random_state=42,init='k-means++')
    model.fit(scaled_data)
    wcss_reduced.append(model.inertia_)
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
# Plot the results
plt.plot(range(1,19) , wcss_reduced,marker='o',linestyle='--')
plt.title('K Means Clustering Elbow Curve to determine ')
plt.xlabel('no. of Clusters')
plt.ylabel('WCSS')
plt.grid(True)
plt.figure(figsize=(12, 8))
plt.show()
```

```python
n_clusters =3
#fit the model
model = KMeans(n_clusters,random_state=42)
model.fit(scaled_data)
#predict
dt['furnishing_type'] = model.predict(scaled_data)
dt.sample(5)[['furnishDetails','furnishing_type']]
# 0 -> unfurnished
# 1 -> semifurnished
# 2 -> furnished
```

###### luxury score

```python
from sklearn.preprocessing import MultiLabelBinarizer
import ast
```

```python
# Convert the string representation of lists in the 'features' column to actual lists
dt['feature_list'] = dt['features'].apply(lambda x: ast.literal_eval(x) if pd.notnull(x) and x.startswith('[') else [])
#extract different values
#use multilabelbinarizer to convert list into binarymatrix
mlb = MultiLabelBinarizer()
feature_binary_matrix = mlb.fit_transform(dt['feature_list'])
#convert to dataframe
features_binary_df = pd.DataFrame(feature_binary_matrix, columns=mlb.classes_)
```

```python
features_binary_df.columns
```

```python
# Define the weights for each feature as provided
# Assigning weights based on perceived luxury contribution
weights = {
    'Airy Rooms': 8,
    'Centrally Air Conditioned': 9,
    'Club house / Community Center': 9,
    'False Ceiling Lighting': 6,
    'Feng Shui / Vaastu Compliant': 5,
    'Fitness Centre / GYM': 8,
    'High Ceiling Height': 8,
    'Intercom Facility': 7,
    'Internet/wi-fi connectivity': 7,
    'Low Density Society': 7,
    'Maintenance Staff': 6,
    'Natural Light': 8,
    'Park': 8,
    'Private Garden / Terrace': 9,
    'Rain Water Harvesting': 7,
    'Security / Fire Alarm': 9,
    'Security Personnel': 9,
    'Separate entry for servant room': 8,
    'Shopping Centre': 7,
    'Spacious Interiors': 9,
    'Swimming Pool': 8,
    'Visitor Parking': 7,
    'Water Storage': 7,
    'Water purifier': 7
}
# Calculate luxury score for each row
luxury_score = features_binary_df[list(weights.keys())].multiply(list(weights.values())).sum(axis=1)
dt['luxury score'] = luxury_score
```
