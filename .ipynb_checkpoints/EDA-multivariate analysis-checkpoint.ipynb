### DataSet import

```python
data = pd.read_csv("Dataset/gurgaon_properties_cleaned.csv")
```

```python
dt = data.copy()
```

```python
dt.columns
```

### property_type 

```python
dt.groupby("property_type")['price'].agg(['min','max','mean','std','count','quantile'])
```

```python
k1 = dt.iloc[dt.groupby("property_type")['price'].groups['flat']]
k1.head(10)

k2 = dt.iloc[dt.groupby("property_type")['price'].groups['house']]
k2.head(10)
```

```python
plt.figure(figsize=(15,6))
plt.subplot(1,2,1)
sns.histplot(k1['price'],color='lightgreen')
plt.title('price of Flats')
plt.grid()

plt.subplot(1,2,2)
sns.histplot(k2['price'],color='lightgreen')
plt.title('price of Houses')
plt.grid()
plt.show()
```

```python
plt.figure(figsize=(15,6))
plt.subplot(1,2,1)
sns.histplot(np.log(k1['price']),color='lightgreen')
plt.title('price of Flats')
plt.grid()

plt.subplot(1,2,2)
sns.histplot(np.log(k2['price']),color='lightgreen')
plt.title('price of Houses')
plt.grid()
plt.show()
```

```python
k1['price'].skew(),k2['price'].kurtosis(),np.log(k1['price']).skew(),np.log(k1['price']).kurtosis(),np.log(k2['price']).skew() , np.log(k2['price']).kurtosis()
```

```python
sns.boxplot(x=dt['property_type'],y=dt['price'],color='lightgreen')
```

```python
sns.barplot(x=dt['property_type'],y=dt['price'],color='lightgreen')
```

```python
k1['price'].cumsum().loc[3692],k2['price'].cumsum().loc[984]
```

##### Remarks:

1. price as per flat data is skewed - right skewed
2. by applying log transformation - the distribution becomes normal around 0
3. too many outliers
4. average price of all plots is like 1.7 , house is like 4.9

```python
plt.figure(figsize=(10,50))
k =  dt.groupby('sector')['price'].agg(['mean'])

sns.heatmap(k,annot=True,linewidth=0.2,fmt = '.2f', annot_kws={'size': 8})
plt.title('Average Price per Sector (Sorted by Sector Number)')
plt.xlabel('Average Price')
plt.ylabel('Sector')
```

```python
plt.figure(figsize=(10,50))
k =  dt.groupby('sector')['luxury score'].agg(['mean'])

sns.heatmap(k,annot=True,linewidth=0.2,fmt = '.2f', annot_kws={'size': 8})
plt.title('Average luxury score per Sector (Sorted by Sector Number)')
plt.xlabel('Average luxury score')
plt.ylabel('Sector')
```

### correlation

```python
num=[]
for i in dt.columns:
    if dt[i].dtype !='object':
        num.append(i)
k = dt[num]    
plt.figure(figsize=(8,8))
sns.heatmap(k.corr())
```
