```python
import requests
# !pip install selenium 
from selenium import webdriver
from selenium.webdriver.edge.service import Service
```

```python
import pandas as pd
import numpy as np 
from bs4 import BeautifulSoup
import lxml 
#xml is a powerful and Pythonic library for processing XML and HTML documents, 
#combining the speed of C libraries with a simple Python interface.
```

```python
header = {'User-Agent':'Mozilla/5.0 (Windows NT 6.3; Win 64 ; x64) '}
webpage = requests.get('https://www.99acres.com/search/property/buy/gurgaon?city=8&keyword=gurgaon&preference=S&res_com=R&page=1',headers=header).text
soup = BeautifulSoup(webpage,'lxml')
```

```python
s = soup.prettify()
```

```python
soup.find_all('div',class_ ='tupleNew__innerCont')
```

```python

for page in range(1,180):
    url = f'https://www.99acres.com/search/property/buy/gurgaon?city=8&keyword=gurgaon&preference=S&res_com=R&page={page}'
    response = requests.get(url,headers= header).text
    soup = BeautifulSoup(response, "html.parser")
    print(soup.find_all('div',class_ ='tupleNew__innerCont'))
```
