# **DataSet Import**

```python
import pandas as pd
import numpy as np
```

```python
data  = pd.read_csv("Dataset/flats.csv")
data.sample(10)
```

```python
# rename columns
data.rename(columns={'area':'price_per_sqft'},inplace=True)
data = data.drop(columns=['link','property_id'])
```

```python
dt= data.copy()
```

# **Cleaning**

```python
dt.head(3)
```

```python
dt.shape
```

**Price**

```python
pd.set_option('display.max_rows',10) # to print complete output rather than some lines
dt['price'].value_counts()
```

```python
# delete rows - since ratio of rows:col is very high ,
# therefore can delete some rows and as price is very imp factor even which will be taken as output therefore can't mess with it 
indexes_to_delete = (dt[dt['price']=='price'].index.tolist()) + (dt[dt['price']=='Price on Request'].index.tolist()) + (dt[dt['price'].isnull()].index.tolist())
print(indexes_to_delete)
dt.drop(indexes_to_delete,axis=0,inplace=True,errors='ignore')
dt.shape
```

```python
pd.set_option('display.max_rows',None)
dt['price'] = dt['price'].str.split(" ").apply( lambda x: round(float(x[0]),2) if x[1]=='Crore' else round(float(x[0])*0.01,3) )
```

```python
dt.head(3)
```

price_per_sqft

```python
dt[dt['price_per_sqft'].isnull()] # ->no null 
```

```python
k = dt['price_per_sqft'].str.split("/").str[0].str.split(' ').str[-1].str.replace(',','').apply(lambda x: float(x))
dt['price_per_sqft'] = k
```

bedroom , bathroom , balcony

```python
print(dt['bedRoom'].value_counts())
print(dt['bedRoom'].isnull().sum())
```

```python
import re
k = dt['bedRoom'].apply(lambda x: int(re.search('\d',x).group()))
dt['bedRoom'] =k
```

```python
print(dt['bathroom'].value_counts())
print(dt['bathroom'].isnull().sum())
```

```python
k = dt['bathroom'].apply(lambda x: int(re.search('\d',x).group()))
dt['bathroom'] =k
```

```python
print(dt['balcony'].value_counts())
print(dt['balcony'].isnull().sum())
```

```python
k = dt['balcony'].apply(lambda x: x.split(" ")[0])
dt['balcony'] =k
```

additionalRoom

```python

dt['additionalRoom'].value_counts()
```

```python
dt['additionalRoom'] = dt['additionalRoom'].fillna('Not Avaliable')
```

```python
dt.head(4)
```

facing , floorNum

```python
print(dt['facing'].isnull().sum())
print(dt['facing'].value_counts())
```

```python
dt['facing'] = dt['facing'].fillna('Not Avaliable')
```

```python
dt[dt['floorNum'].isnull()]
dt['floorNum'] = dt['floorNum'].fillna('6')
```

```python
pd.set_option('display.max_rows',None)
```

```python
dt['floorNum'].value_counts()
```

```python
dt['floorNum'] = dt['floorNum'].apply(lambda x: re.search('\d+',x).group())
```

```python
dt['area'] = round(dt['price']*10000000/dt['price_per_sqft'],3)
```

```python
col = dt.pop('area')
dt.insert(2, 'area', col)
dt.head(10)
```

```python
dt.info()
```

```python
# check for duplicates
dt.duplicated().sum()
dt = dt.drop_duplicates()
```

```python
dt.loc[(dt['society']=='nul') | (dt['society']=='Not Applicable'),'society'] = dt.loc[(dt['society']=='nul') | (dt['society']=='Not Applicable')]['property_name'].str.split("in").str[1]
```

```python
dt['society'].isnull().sum()
```

```python
dt.insert(loc=1,column='property_type',value='flat')
```

# **Export**

```python
dt.to_csv("Dataset/flats-basic cleaned.csv",index=False)
print("successfull")
```
