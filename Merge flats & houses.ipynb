# **Merge Flats and Houses**

```python
import pandas as pd
import numpy as np
```

```python
flats = pd.read_csv("Dataset/flats-basic cleaned.csv")
houses = pd.read_csv("Dataset/houses-basic cleaned.csv")
```

```python
flats.columns
```

```python
houses.columns
```

```python
k = pd.concat([flats,houses])
```

```python
data = k.sample(k.shape[0])
```

```python
data.shape
```

```python
data.info()
```

```python
data['sector'] = data['property_name'].str.split('in').str[1].str.strip()
```

```python
c = data.pop('sector')
data.insert(3,'sector',c)
```

# Export

```python
data.to_csv('Dataset/gurgaon-properties.csv',index=False)
print("successfull")
```
