```python
dt.shape
```

price , area , price_per_sqft 

```python

print(dt['price'].isnull().sum(),dt['price_per_sqft'].isnull().sum(),dt['area'].isnull().sum())
#can fill total price - rate*area
```

```python
list = dt[(dt['price_per_sqft'].isnull()) & (dt['price'].isnull())].index.tolist()
dt = dt.drop(list,axis=0)
```

```python
def transform(x):
    p = x.split("-")
    if len(p)==1:
        return int(x)*10.7639
    else :
        a = int(p[0])
        b = int(p[1])
        c = (a+b)/2
        return c*10.7639
transform("123-321")
```

```python

dt['area'] = dt['area'].str.split("sq.m.").str[0].str.replace("(","").apply(lambda x:transform(x)).astype('float')
```

```python
pd.set_option("display.max_rows",None)
```

```python
#rate
dt['price_per_sqft'] = dt['price_per_sqft'].fillna('₹ 0,0/sq.ft.')
dt['price_per_sqft'] = dt['price_per_sqft'].str.split("/").str[0].str.split(" ").str[1].str.replace(",","").astype('float')
```

```python
dt.loc[dt['price'].isnull(),'price'] = dt.loc[dt['price'].isnull(),'area']*dt.loc[dt['price'].isnull(),'price_per_sqft']
dt = dt.drop(dt[dt['price'] =='Price on Request'].index.tolist())
```

```python
def transform(x):
    x = str(x)
    if x.find('Crore')!=-1 :
        num = float(x.split(' ')[0])
        return num
    elif x.find('Lac')!=-1:
        num = float(x.split(' ')[0])*0.01
        return num
    else:
        num = float(x)
        return num*0.0000001
        
        
```

```python
dt['price'] = dt['price'].apply(lambda x:round(transform(x),3))
```
